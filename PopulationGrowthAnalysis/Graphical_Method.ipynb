# **Comparative graphical method**
## **Example**: The population of a city A as per the censuses records are given below. Estimate the population of city A in year 2040 and 2050 based on the records for cities B, C, and D.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
yiA = [1990,2000,2010,2020]
PiA = [32000,38000,43000,55000]
maxPiA = np.max(PiA)
maxYiA = np.max(yiA)
```

```python
yiB = [1980,1992,2000,2010]
PiB = [50000,62000,72000,87000]
```

```python
yiC = [1982,1990,2001,2008]
PiC = [50000,58000,69000,76000]
```

```python
yiD = [1981,1993,2002,2009]
PiD = [50000,54000,58000,62000]
```

```python
# Plot the population data
# plt.plot(yiA, PiA, 'r-', yiB, PiB, 'bs', yiC, PiC, 'g^',yiC,PiD,'ok')
plt.plot(yiA, PiA, 'or-', mfc='none', label='city A')
plt.plot(yiB, PiB, '-bs', label='city B')
plt.plot(yiC, PiC, '-g^', label='city C')
plt.plot(yiC,PiD,'-o', color="0.5", label='city D')
plt.xlabel('Years', fontweight='bold')
plt.ylabel('Population', fontweight='bold')
plt.legend()
plt.show()
```

```python
yiAll = yiB+yiC+yiD
yiAll.sort()
yiAll = np.unique(yiAll)
maxyi = np.max(yiAll)
minyi = np.min(yiAll)
# Linspace vector for yiAll
fullYi = np.arange(minyi,maxyi+5,5)
```

```python
# Interpolate or extrapolate for each year of
for x in fullYi:
  PfB = np.interp(fullYi,yiB,PiB)
  PfC = np.interp(fullYi,yiC,PiC)
  PfD = np.interp(fullYi,yiD,PiD)

# print(PfB)
# print(PfC)
# print(PfD)
```

```python
# GEt the average population data from cites B, C and D
PfB = np.around(PfB)
PfC = np.around(PfC)
PfD = np.around(PfD)

aveArr = (PfB+PfC+PfD)/3
aveArr = np.around(aveArr)
minaveArr = np.min(aveArr)
```

```python
# Interpolate to translate the average population (black dashed lines)
PiA2transl = np.interp(minaveArr,PiA,yiA)

# Line of min average
yiAllMin = yiA+yiB+yiC+yiD
yiAllMin.sort()
yiAllMin = np.unique(yiAllMin)
maxyiMin = np.max(yiAllMin)
minyiMin = np.min(yiAllMin)

# Linspace vector for dotted line
fullYiMin = np.arange(minyiMin,maxyiMin+5,5)
leng = len(fullYiMin)
PMinOfAve = minaveArr * np.ones(leng)
```

```python
# Plot the average population data
plt.plot(yiA, PiA, 'or-', mfc='none', label='city A')
plt.plot(yiB, PiB, '-b', label='city B')
plt.plot(yiC, PiC, '-g', label='city C')
plt.plot(yiC,PiD,'-', color="0.5", label='city D')

plt.plot(fullYi,aveArr,'--sk', label='Mean')

#plt.plot(fullYiMin,PMinOfAve,':',color="0.5")

#plt.plot(PiA2transl,minaveArr,'rs')

plt.xlabel('Years', fontweight='bold')
plt.ylabel('Population', fontweight='bold')
plt.legend()
plt.show()
```

```python
# To translate the average line
deltaY = maxYiA - minyiMin
deltaP = maxPiA - minaveArr

newY4Averag = np.ones(len(fullYi))*deltaY + fullYi
newP4Averag = np.ones(len(fullYi))*deltaP + aveArr
```

```python
# Plot translated average line from the intersection on
plt.plot(yiA, PiA, 'or-', mfc='none', label='city A')
plt.plot(yiB, PiB, '-b', label='city B')
plt.plot(yiC, PiC, '-g', label='city C')
plt.plot(yiC, PiD,'-', color="0.5", label='city D')

plt.plot(newY4Averag,newP4Averag,'ks-', label='added to city A')

plt.xlabel('Years', fontweight='bold')
plt.ylabel('Population', fontweight='bold')
plt.legend()
plt.grid()
plt.show()
```
